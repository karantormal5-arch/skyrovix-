```python
import pandas as pd

SRC = "Online_Retail.csv"
OUT = "OnlineRetail_clean.csv"

df = pd.read_csv(SRC, encoding="utf-8-sig", dtype={"InvoiceNo": str, "StockCode": str})
print("Raw:", len(df))

df["Description"] = df["Description"].str.strip().str.upper()
df["Country"] = df["Country"].str.strip()

df = df.drop_duplicates()
print("After duplicates:", len(df))

df = df.dropna(subset=["Description", "CustomerID"])
print("After blank desc/customer:", len(df))
df["CustomerID"] = df["CustomerID"].astype(int).astype(str)

df = df[~df["InvoiceNo"].str.startswith("C")]
print("After cancellations:", len(df))

df = df[(df["Quantity"] > 0) & (df["UnitPrice"] > 0)]
print("After qty/price:", len(df))

df = df[~df["StockCode"].isin(["POST", "M", "C2", "DOT", "BANK CHARGES", "PADS"])]
print("After non-product:", len(df))

df["InvoiceDate"] = pd.to_datetime(df["InvoiceDate"], format="%m/%d/%y %H:%M").dt.normalize()
df["Revenue"] = (df["Quantity"] * df["UnitPrice"]).round(2)
df["InvoiceMonth"] = df["InvoiceDate"].dt.to_period("M").astype(str)

print("Invoices:", df["InvoiceNo"].nunique())
print("Customers:", df["CustomerID"].nunique())
print("Total revenue:", round(df["Revenue"].sum(), 2))

df.to_csv(OUT, index=False, encoding="utf-8")
print("Saved:", OUT)
```
